# Bible Gateway Reference Lookup

This optional notebook demonstrates how to retrieve cross-reference rows for one source verse and attach Bible Gateway links.

```python
from pathlib import Path
import sys

import pandas as pd

cwd = Path.cwd().resolve()
if (cwd / "data").exists() and (cwd / "src").exists():
    PROJECT_DIR = cwd
elif cwd.name == "notebooks" and (cwd.parent / "data").exists():
    PROJECT_DIR = cwd.parent
elif (cwd / "revelation" / "data").exists():
    PROJECT_DIR = cwd / "revelation"
else:
    raise FileNotFoundError("Could not locate the revelation/data directory from the current working directory.")

SRC_DIR = PROJECT_DIR / "src"
DATA_DIR = PROJECT_DIR / "data"

if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

from biblegateway_links import add_biblegateway_links, get_source_verse_references, make_biblegateway_url
from openbible_data import load_cross_references
```

```python
df = load_cross_references(DATA_DIR / "cross_references_revelation_public.csv")
df.shape
```

## Lookup Parameters

Change `book`, `chapter`, `verse`, or `version` to inspect another source verse.

```python
book = "Rev"
chapter = 22
verse = 1
version = "ESV"

source_reference = f"{book}.{chapter}.{verse}"
source_url = make_biblegateway_url(source_reference, version=version)
source_url
```

```python
references = get_source_verse_references(df, book=book, chapter=chapter, verse=verse)
references_with_links = add_biblegateway_links(references, reference_column="To Verse", version=version)

references_with_links[[
    "From Verse",
    "To Verse",
    "To Verse start Book",
    "To Verse start Chapter",
    "To Verse start number",
    "Bible Gateway URL",
]]
```
